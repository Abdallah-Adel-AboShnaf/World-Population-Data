```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
numeric_cols = df.select_dtypes(include = np.number).columns.tolist()
if 'Rank' in numeric_cols:
    numeric_cols.remove('Rank')
numeric_cols
```

```python
import seaborn as sns
for col in numeric_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
for column in numeric_cols:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    rows_before = df.shape[0]

    df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

    rows_after = df.shape[0]
    print(f"Column '{column}': {rows_before - rows_after} rows removed as outliers.")

print("\n-------------------------------------------")
print(f"Original data size: {df.shape}")
print(f"Data size after removing outliers: {df.shape}")
```
